```python
import pandas as pd
import numpy as np
df = pd.read_excel("online_retail_cleaned.xlsx")
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
df
```

##### Total Net Revenue

```python
print(f" Total Net Revenue: {df["Revenue"].sum()}")
```

#### Year Wise Revenue Sum

```python
year = df["InvoiceDate"].dt.year
df.groupby(year)["Revenue"].sum()
```

#### Month Wise Revene Sum

```python
month = df['InvoiceDate'].dt.month
df.groupby(month)['Revenue'].sum()
```

#### Year and Month Wise Revenue Sum

```python
year_month = df['InvoiceDate'].dt.strftime('%Y-%m')
year_month_revenue = df.groupby(year_month)['Revenue'].sum().reset_index()
year_month_revenue
```

#### Country Revenue Sum

```python
crs = df.groupby("Country")["Revenue"].sum()
crs
```

#### TOP 10 Countries

```python
crs.sort_values(ascending=False).head(10)
```

#### Revenue Sum according to Invoice Number

```python
inv = df.groupby("InvoiceNo")["Revenue"].sum()
inv
```

#### Top 10 Invoices

```python
inv.sort_values(ascending=False).head(10)
```

#### Average Invoice

```python
inv.mean()
```

#### Average product lines per invoice

```python
total = df["InvoiceNo"].count()
uniq = df["InvoiceNo"].nunique()
avg = total / uniq
avg
```

#### Revenue Reconciliation

```python
rev_sum = df["Revenue"].sum()
inv_sum = inv.sum()
if np.isclose(rev_sum,inv_sum):
    print("Both values are equal")
else:
    print("Both Values are not equal")
```
