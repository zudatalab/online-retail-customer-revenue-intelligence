```python
import pandas as pd
import numpy as np
df = pd.read_excel("online_retail_cleaned.xlsx")
```

#### Top 10 Customers

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
cus = df.groupby('CustomerID')['Revenue'].sum()
cus.sort_values(ascending=False).head(10).reset_index()
```

#### Every Customer Unique Invoice

```python
cui = df.groupby('CustomerID')['InvoiceNo'].value_counts().reset_index()
cui
```

#### Top 10 Invoice Count Customers

```python
df.groupby('CustomerID')['InvoiceNo'].nunique().sort_values(ascending=False).head(10).reset_index()
```

#### Every Customer average invoice value

```python
cir = df.groupby(["CustomerID","InvoiceNo"])["Revenue"].sum()
result = cir.groupby("CustomerID").mean().reset_index()
result
```

#### Top 10 of customer average invoice value

```python
result.sort_values(by="Revenue", ascending=False).head(10)
```

#### Customer Total Purchase Quantity

```python
total = df.groupby("CustomerID")["Quantity"].sum().reset_index()
total
```

##### Top 10 (Customer Total Purchase Quantity)

```python
total.sort_values(by="Quantity",ascending=False).head(10)
```

#### How many unique products has each customer purchased?

```python
unqi_stock = df.groupby("CustomerID")["StockCode"].nunique().reset_index()
unqi_stock
```

##### Top 10

```python
unqi_stock.sort_values(by="StockCode",ascending=False).head(10)
```

#### Customer Purchase Frequency

```python
freq = df.groupby('CustomerID')['InvoiceNo'].nunique().reset_index()
freq[freq["InvoiceNo"] >= 5]
```

#### Identify the date of the first purchase and the date of the last purchase for each customer.

```python
duration = df.groupby("CustomerID")["InvoiceDate"].agg(First_InvoiceDate="min", Last_InvoiceDate="max").reset_index()
duration
```

```python
whole_days = (duration["Last_InvoiceDate"] - duration["First_InvoiceDate"]).reset_index()
whole_days
```

```python
customer_summary = df.groupby("CustomerID").agg(
    Total_Net_Revenue=("Revenue", "sum"),
    Unique_Invoice_Count=("InvoiceNo", "nunique"), 
    Total_Net_Quantity=("Quantity", "sum"), 
    Unique_StockCode_Count=("StockCode", "nunique"),  
    First_Purchase_Date=("InvoiceDate", "min"),       
    Last_Purchase_Date=("InvoiceDate", "max")         
).reset_index()
customer_summary["Purchase_History_Duration_Days"] = (
    customer_summary["Last_Purchase_Date"] - customer_summary["First_Purchase_Date"]
).dt.days

customer_summary
```

#### Save File 

```python
customer_summary.to_excel("customer_summary.xlsx", index=False)
```
