```python
import pandas as pd
import numpy as np
df = pd.read_excel("online_retail_cleaned.xlsx")
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```

#### Product Revenue Analysis

```python
pra = df.groupby("StockCode")["Revenue"].sum().reset_index()
pra
```

###### Top 10

```python
pra.sort_values(by="Revenue",ascending=False).head(10)
```

#### Product Quantity Analysis

```python
pqa = df.groupby("StockCode")["Quantity"].sum().reset_index()
pqa
```

##### PQA TOP 10

```python
pqa.sort_values(by="Quantity",ascending=False).head(10)
```

#### Country Performance

##### Total Net Revenue

```python
ctnr = df.groupby("Country")["Revenue"].sum().reset_index()
ctnr
```

##### Country Unique Invoices

```python
cui = df.groupby("Country")["InvoiceNo"].nunique().reset_index()
cui
```

##### Country Unique Customer

```python
cuc = df.groupby("Country")["CustomerID"].nunique().reset_index()
cuc
```

```python
country_summary = df.groupby("Country").agg(
    Total_Net_Revenue=("Revenue","sum"),
    Unique_Invoices=("InvoiceNo","nunique"),
    Unique_Customers=("CustomerID","nunique")).reset_index()
country_summary
```

##### TOP 10 COUNTRY

```python
country_summary.sort_values(by="Total_Net_Revenue",ascending=False).head(10)
```

## Business Insights

1. **United Kingdom dominates revenue:** The United Kingdom generated the highest total net revenue of approximately **8.17 million**, making it the company's largest market by a significant margin.

2. **Strong customer and invoice base in the UK:** The United Kingdom recorded **23,494 unique invoices** and **3,950 unique customers**, considerably more than any other country in the dataset. This indicates that the business relies heavily on the UK market.

3. **The Netherlands is the second-largest market:** The Netherlands generated approximately **284,661.54** in net revenue, followed by EIRE at **262,993.38** and Germany at **221,509.47**. These countries represent important markets outside the UK.

4. **Revenue is concentrated in a few countries:** The difference between the UK's revenue and that of the other countries is substantial. This suggests a strong dependence on the UK market and highlights an opportunity to grow sales in other countries.

5. **Product performance should be evaluated using multiple metrics:** Products with the highest net revenue may not always have the highest net quantity. Comparing both metrics helps distinguish high-revenue products from products sold in larger quantities.

6. **Special product codes require further investigation:** The code `DOT` ranked first in the product revenue analysis. Its meaning should be verified before treating it as a regular product, since special codes may represent services or other charges.
