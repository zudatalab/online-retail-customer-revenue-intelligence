## 3. Quantity Integrity Audit

#### Quantity Investigation

```python
print(f"Quantity Minimum Value: {df["Quantity"].min()}")
print(f"Quantity Maximum Value: {df["Quantity"].max()}")
print(f"Number of less then 0 Values: {(df["Quantity"] < 0).sum()}")
print(f"Number of equal to 0 Values: {(df["Quantity"] == 0).sum()}")
print(f"Percentage of less then 0 Values: {((df["Quantity"] < 0).mean() * 100).round(2)}%")
```

#### Basic Comparison Between Positive and Negative Quantity

```python
# Positive Quantity
print(" --- Positive Quantity --- ")
print(f"Total Positive Qunatity Rows: {(df["Quantity"] > 0).sum()}")
print(f"Average of Positive Qunatity Rows: {df.loc[df["Quantity"] > 0, "Quantity"].mean()}")
print(f"Maximum Value of Positive Qunatity Rows: {df[df["Quantity"] > 0]["Quantity"].max()}")
print(f"Minimum Value of Positive Qunatity Rows: {df[df["Quantity"] > 0]["Quantity"].min()}")
# Negative Quntity
print(" --- Negative Quantity --- ")
print(f"Total Negative Qunatity Rows: {(df["Quantity"] < 0).sum()}")
print(f"Average of Negative Qunatity Rows: {df.loc[df["Quantity"] < 0, "Quantity"].mean()}")
print(f"Maximum Value of Negative Qunatity Rows: {df[df["Quantity"] < 0]["Quantity"].max()}")
print(f"Minimum Value of Negative Qunatity Rows: {df[df["Quantity"] < 0]["Quantity"].min()}")
```

### 8. Cross Column Integrity

##### Quantity ↔ UnitPrice

```python
print(f"Number of rows where quantity is positive and unit price is negative or zero: {((df["Quantity"] > 0 ) & (df["UnitPrice"] <= 0)).sum()}")
print(f"This Condition Percentage: {(((df["Quantity"] > 0 ) & (df["UnitPrice"] <= 0)).mean() * 100).round(2)}%")
```

```python
df[(df["Quantity"] > 0 ) & (df["UnitPrice"] <= 0)].head(5)
```

```python
print(f"Number of rows where quantity and Unit Price Both Less then Zero: {((df["Quantity"] < 0 ) & (df["UnitPrice"] < 0)).sum()}")
print(f"This Condition Percentage: {(((df["Quantity"] < 0 ) & (df["UnitPrice"] < 0)).mean() * 100).round(2)}%")
```
